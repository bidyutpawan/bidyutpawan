```python
name = 'Bidyut'
city = "Bangalore"
multi = '''Data Engineer
            at Cognizant'''

#Indexing and Slicing of Strings
print(name[0])   #B - first character of the string
print(name[1])   #i - second character of the string
print(name[-1])  #t - last character of the string
print(name[1:4]) #idy - characters from index 1 to 3
print(name[::-1]) #tuydiB - reverse of the string

```

```python


#commonly used string methods
print(name.upper()) #BIDYUT - converts all characters to uppercase
print(name.lower()) #bidyut - converts all characters to lowercase
print(name.replace('B','b')) #bidyut - replaces 'B' with 'b'
print(name.split('i')) #['B', 'dyut'] - splits the string
print(name.strip()) #Bidyut - removes leading and trailing whitespaces
print(name.find('d')) #2 - returns the index of the first occurrence of 'd'
```
